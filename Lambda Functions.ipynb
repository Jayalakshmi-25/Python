Section 1: Lambda Expressions — 20 Questions

Question 1:Write a lambda expression to calculate the square of number.

```python
number = 8
square = lambda x: x**2
print(square(number))
```

Question 2:Write a lambda expression to calculate the cube of number.

```python
number = 5
cube = lambda x: x**3
print(cube(number))
```

Question 3:Write a lambda expression to add a and b.



```python
a = 25
b = 17
add = lambda a, b: a+b
print(add(a, b))
```

Question 4:Write a lambda expression to subtract b from a.



```python
a = 50
b = 18
subtract = lambda a, b: a-b
print(subtract(a, b))
```

Question 5:Write a lambda expression to multiply a and b.

```python
a = 12
b = 9
multiply = lambda a, b: a*b
print(multiply(a, b))
```

Question 6:Write a lambda expression to divide a by b.

```python
a = 100
b = 4
divide = lambda a, b: a/b
print(divide(a, b))
```

Question 7:Write a lambda expression to return the larger number.

```python
a = 72
b = 91
larger = lambda a, b: max(a, b)
print(larger(a, b))
```

Question 8:Write a lambda expression to return the smaller number.

```python
a = 46
b = 28
smaller = lambda a, b: min(a, b)
print(smaller(a, b))
```

Question 9:Write a lambda expression that returns True when the number is even and False otherwise.

```python
number = 34
even = lambda number: number%2==0
print(even(number))
```

Question 10:Write a lambda expression that returns True when the number is odd and False otherwise.

```python
number = 27
odd = lambda number: number%2!=0
print(odd(number))
```

Question 11:Write a lambda expression to calculate the area of a rectangle.

```python
length = 12
width = 8
rectangle = lambda length, width: length*width
print(rectangle(length, width))
```

Question 12:Write a lambda expression to calculate the area of a circle using 3.14.

```python
radius = 7
circle = lambda radius: 3.14*radius**2
print(circle(radius))
```

Question 13:Write a lambda expression to calculate the perimeter of a rectangle.

```python
length = 15
width = 10
rectangle = lambda length, width: 2*(length+width)
print(rectangle(length, width))
```

Question 14:Write a lambda expression to convert Celsius into Fahrenheit.

```python
celsius = 35
fahrenheit = lambda celsius: (celsius*9/5)+32
print(fahrenheit(celsius))
```

Question 15:Write a lambda expression to convert Fahrenheit into Celsius.


```python
fahrenheit = 104
celsius = lambda fahrenheit: (fahrenheit-32)*5/9
print(celsius(fahrenheit))
```

Question 16:Write a lambda expression to return the length of the string.

Question 18:Write a lambda expression to return the last character of the string.

```python
word = "Databricks"
last = lambda word: word[-1]
print(last(word))
```

Question 19:Write a lambda expression to return the absolute value of the number without using abs().



```python
number = -45
absolute = lambda number: number*(-1) if number<0 else number
print(absolute(number))
```

Question 20:Write a lambda expression that returns:

"Distinction" when marks are 75 or above
"Pass" when marks are 40 to 74
"Fail" when marks are below 40

```python
marks = 72
result = lambda marks: "Distinction" if marks>=75 else "Pass" if marks>=40 else "Fail"
print(result(marks))
```
