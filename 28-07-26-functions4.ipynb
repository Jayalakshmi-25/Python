**Assignment 35: Fibonacci Function**

Write a function named generate_fibonacci().
Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.
Example Input
Enter number of terms: 7
Expected Output
[0, 1, 1, 2, 3, 5, 8]

```python
def generate_fibonacci(number_of_terms):
    if number_of_terms <= 0:
        return "Invalid number of terms."
    fibonacci_series = [0, 1]
    while len(fibonacci_series) < number_of_terms:
        next_number = fibonacci_series[-1] + fibonacci_series[-2]
        fibonacci_series.append(next_number)
    return fibonacci_series
number_of_terms = int(input("Enter number of terms: "))
fibonacci_series = generate_fibonacci(number_of_terms)
print("Fibonacci Series:", fibonacci_series)
```

```python
def create_bank_account():
    balance = 0
    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance
    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            return "Insufficient balance"
        balance -= amount
        return balance
    def check_balance():
        return balance
    return deposit, withdraw, check_balance
deposit, withdraw, check_balance = create_bank_account()
print(deposit(100))
print(withdraw(50))
print(check_balance())
```

**Assignment 93: Function Decorator**

It should print another message after execution.
Apply it to an addition function.
Preserve the original function's return value.
Expected Pattern
Starting function...
Result: 30
Function completed.

```python
def execution_logger(func):
    def wrapper(*args, **kwargs):
        print("Starting function...")
        result = func(*args, **kwargs)
        print("Function completed.")
        return result
    return wrapper
@execution_logger
def add(a, b):
    return a + b
result = add(10, 20)
print("Result:", result)
```
